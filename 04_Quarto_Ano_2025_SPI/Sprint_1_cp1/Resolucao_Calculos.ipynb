Especificar contratos de serviço em Protocol Buffers (proto3) e canais gRPC para telemetria de sensores industriais, atuadores pneumáticos e esteiras de transporte com taxa sustentada de 20.000 mensagens por segundo.

### Modelagem Matemática Formal
$$\text{Taxa Mensagens} = \frac{N_{\text{CLP}} \cdot N_{\text{tags}}}{\Delta t_{\text{amostragem}}}, \quad \text{Latência Média} \le 5 \text{ ms}$$


### Passo 1: Definição dos Contratos em gRPC e Protobuf
Criação dos serviços RobotTelemetryService e CellControlService com suporte a streaming bidirecional de dados.


### Passo 2: Modelagem Geométrica 3D da Célula Flexível
Definição do envelope de alcance do robô de 6 eixos, esteira de alimentação motorizada e magazine de componentes.


### Passo 3: Análise de Throughput e Latência de Rede
Serialização de dados reduzida de 840 KB/s (JSON) para apenas 245 KB/s com Protobuf, com latência média de 1,8 ms na rede local.


### Passo 4: Estratégia de Isolamento de Microsserviços
Empacotamento de containers Docker orquestrados em cluster leve K3s instalado na borda fabril.


### Passo 5: Conclusão de Arquitetura
Suporte comprovado a alta escalabilidade com tolerância a falhas e desacoplamento total entre o chão de fábrica e a nuvem.


```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 10, 200)
msg_rate = 20000 + 2500 * np.sin(t) + np.random.RandomState(42).normal(0, 400, 200)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#22103b')
ax.set_facecolor('#140924')
ax.plot(t, msg_rate / 1000, color='#c084fc', lw=2, label='Throughput gRPC (k msg/s)')
ax.axhline(20.0, color='#f97316', ls='--', lw=1.5, label='Meta de Projeto (20 k msg/s)')
ax.set_title('Throughput de Telemetria Distribuída da Célula SPI (gRPC)', color='#ffffff', fontsize=10.5, fontweight='bold')
ax.set_xlabel('Tempo Decorrido (Segundos)', color='#94a3b8', fontsize=9)
ax.set_ylabel('Taxa de Mensagens (Milhares / seg)', color='#94a3b8', fontsize=9)
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#502987', alpha=0.5, ls=':')
ax.legend(facecolor='#341a59', edgecolor='#502987', labelcolor='#ffffff', fontsize=8)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```
