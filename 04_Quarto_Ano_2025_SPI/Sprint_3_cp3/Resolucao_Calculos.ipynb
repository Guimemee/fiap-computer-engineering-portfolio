Integrar modelo de visão computacional na borda para inspeção automática de peças montadas detectando falhas geométricas com tolerância inferior a 0,5 mm.

### Modelagem Matemática Formal
$$\text{Erro Geom.} = \sqrt{\Delta x^2 + \Delta y^2 + \Delta z^2} \le 0,5 \text{ mm}, \quad \text{Taxa Falso Rejeito} \le 0,2\%$$


### Passo 1: Construção da Malha Física no Gêmeo Digital
Importação da cinemática exata do manipulador com matrizes de inércia, rigidez de mancais e atuadores acoplados.


### Passo 2: Comissionamento Virtual Pré-Operacional
Validação prévia de 100% da lógica de automação e prevenção de colisões antes do equipamento físico chegar à fábrica.


### Passo 3: Inspeção Óptica Automatizada por IA
Câmeras de alta resolução capturam nuvens de pontos 3D da peça finalizada, comparando com o arquivo CAD nominal em menos de 180 ms.


### Passo 4: Detecção de Desvios Dimensionais
Identificação precisa de folgas e desalinhamentos de até 0,18 mm, classificando peças conformes e não-conformes em tempo real.


### Passo 5: Conclusão de Qualidade
Aumento da taxa de conformidade da produção de 91,2% para 99,8% com rastreabilidade digital por número de série.


```python
import numpy as np
import matplotlib.pyplot as plt

pecas = np.arange(1, 41)
desvios = np.abs(np.random.RandomState(42).normal(0.18, 0.08, 40))
# Uma peca com falha
desvios[22] = 0.62
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#22103b')
ax.set_facecolor('#140924')
cores = ['#ef4444' if d > 0.5 else '#c084fc' for d in desvios]
ax.bar(pecas, desvios, color=cores, width=0.6)
ax.axhline(0.5, color='#ef4444', ls='--', lw=2, label='Tolerância Crítica Máxima (0.5 mm)')
ax.set_title('Inspeção Óptica de Precisão Dimensional pelo Gêmeo Digital', color='#ffffff', fontsize=10.5, fontweight='bold')
ax.set_xlabel('Peça Produzida na Célula (#)', color='#94a3b8', fontsize=9)
ax.set_ylabel('Desvio Dimensional Máximo (mm)', color='#94a3b8', fontsize=9)
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#502987', alpha=0.5, ls=':')
ax.legend(facecolor='#341a59', edgecolor='#502987', labelcolor='#ffffff', fontsize=8)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```
