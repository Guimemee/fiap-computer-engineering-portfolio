> Projetar a arquitetura eletrônica de potência e sensoriamento de desvio de obstáculos para navegação em corredores hospitalares com fluxo de pessoas. Dimensionar o banco de baterias de Lítio Ferro Fosfato (LiFePO4) de 24 V para sustentar autonomia mínima de 12 horas contínuas de operação considerando potência média consumida de 95 W e profundidade máxima de descarga (DoD) de 80%.

### Modelagem Matemática Formal
$$E_{\text{requerida}} = \frac{P_{\text{médio}} \cdot t}{\text{DoD} \cdot \eta_{\text{conv}}}, \quad C_{\text{Ah}} = \frac{E_{\text{Wh}}}{V_{\text{nom}}}$$


### Passo 1: Energia Útil Demandada em 12 Horas
Energia pura consumida: E = 95\text{ W} \times 12\text{ h} = 1.140\text{ Wh}.


### Passo 2: Consideração de Perdas e Profundidade de Descarga
Com rendimento dos conversores DC-DC \eta = 92% e DoD seguro de 80% (0,80) para garantir vida útil de 3.000 ciclos: E_{nominal} = \frac{1.140}{0,80 \times 0,92} = \frac{1.140}{0,736} \approx 1.548,9\text{ Wh}.


### Passo 3: Capacidade do Banco de Baterias em Amperes-hora
Para tensão nominal de barramento V = 24 V: C = \frac{1.548,9}{24} \approx 64,54\text{ Ah}.


### Passo 4: Topologia dos Módulos LiFePO4
Arranjo de células cilíndricas 32700 de 3,2 V e 6.000 mAh em configuração 8S11P (8 células em série \times 11 em paralelo = 88 células no total), entregando 25,6 V nominais e 66 Ah (1.689,6 Wh).


### Passo 5: Conclusão de Autonomia e Segurança
A configuração atinge 13,8 horas de operação contínua ininterrupta com sistema de proteção BMS (Battery Management System) com monitoramento térmico ativo por leito.


```python
import numpy as np
import matplotlib.pyplot as plt

horas = np.linspace(2, 16, 100)
p_med = 95.0
v_nom = 24.0
dod = 0.8
eta = 0.92
cap_ah = (p_med * horas) / (dod * eta * v_nom)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#0c233c')
ax.set_facecolor('#081626')
ax.plot(horas, cap_ah, color='#38bdf8', lw=2.5, label='Capacidade Necessária (Ah)')
ax.axvline(12, color='#f59e0b', ls='--', label='Meta Hospitalar (12h -> 64.5 Ah)')
ax.axhline(66, color='#10b981', ls=':', lw=2, label='Banco Selecionado 8S11P (66 Ah)')
ax.set_title('Dimensionamento do Banco de Bateria LiFePO4 (24V)', color='#ffffff', fontsize=10.5, fontweight='bold')
ax.set_xlabel('Horas de Operação Ininterrupta', color='#94a3b8', fontsize=9)
ax.set_ylabel('Capacidade da Bateria (Ah)', color='#94a3b8', fontsize=9)
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#1e4d82', alpha=0.5, ls=':')
ax.legend(facecolor='#14355a', edgecolor='#1e4d82', labelcolor='#ffffff', fontsize=8)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```
