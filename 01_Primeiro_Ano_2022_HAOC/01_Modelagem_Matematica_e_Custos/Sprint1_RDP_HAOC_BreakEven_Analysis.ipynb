### 📋 Contexto do Desafio
Nossa startup de engenharia foi contratada para desenvolver uma frota de **robôs autônomos para vistoria hospitalar e interação humanizada com pacientes** no Hospital Alemão Oswaldo Cruz.

Os parâmetros financeiros levantados pela equipe de engenharia para a versão beta são:
- **Custo Unitário de Produção ($C_u$):** R$ 535,00 por robô.
- **Custo Fixo Mensal ($C_F$):** R$ 46.750,00 (aluguel, P&D, infraestrutura, energia e salários técnicos).
- **Preço de Venda Unitário ($P_v$):** R$ 785,00 por robô entregue ao complexo hospitalar.

Todas as funções econômicas são modeladas como polinômios do 1º grau em função de $N$ (unidades produzidas e vendidas).

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

# Declaração simbólica
N = sp.Symbol('N', positive=True, integer=True)
Cu = 535
CF = 46750
Pv = 785
print('Configuração do ambiente concluída com SymPy e Matplotlib!')
```

---
### ❓ Questão 1: Modelo Matemático do Custo Mensal Total $C(N)$
O Custo Total de uma operação fabril é a soma do Custo Fixo com o Custo Variável total:
$$C(N) = C_F + C_u \cdot N$$
Substituindo os valores:
$$C(N) = 46.750 + 535N$$

```python
C = CF + Cu * N
sp.Eq(sp.Symbol('C(N)'), C)
```

---
### ❓ Questão 2: Modelo Matemático da Receita Mensal Total $R(N)$
A receita obtida com a comercialização de $N$ unidades vendidas a $R\$ 785,00$ é dada por:
$$R(N) = P_v \cdot N$$
$$R(N) = 785N$$

```python
R = Pv * N
sp.Eq(sp.Symbol('R(N)'), R)
```

---
### ❓ Questão 3: Modelo Matemático do Lucro Mensal Total $L(N)$
O lucro operacional corresponde à diferença entre a receita e os custos totais:
$$L(N) = R(N) - C(N)$$
$$L(N) = 785N - (535N + 46.750)$$
$$L(N) = (785 - 535)N - 46.750 = 250N - 46.750$$
A **Margem de Contribuição Unitária** é de **R$ 250,00** por robô produzido.

```python
L = sp.simplify(R - C)
sp.Eq(sp.Symbol('L(N)'), L)
```

---
### ❓ Questão 4: Ponto de Equilíbrio (Break-Even Point) e Análise Gráfica
O Ponto de Equilíbrio ocorre no momento em que a Receita Total empata com o Custo Total ($L(N) = 0$):
$$R(N) = C(N) \implies 250N - 46.750 = 0 \implies N^* = \frac{46.750}{250} = 187 \text{ robôs}$$

Financeiramente, o faturamento de equilíbrio é:
$$R(187) = 187 \times 785 = R\$ 146.795,00$$

```python
N_breakeven = sp.solve(L, N)[0]
valor_breakeven = R.subs(N, N_breakeven)
print(f'Ponto de Equilíbrio Operacional:')
print(f'Quantidade N* = {N_breakeven} robôs')
print(f'Receita / Custo no equilíbrio = R$ {valor_breakeven:,.2f}')
```

```python
# Visualização Computacional de Alta Performance
n_vals = np.linspace(0, 350, 400)
c_vals = [float(C.subs(N, val)) for val in n_vals]
r_vals = [float(R.subs(N, val)) for val in n_vals]
l_vals = [float(L.subs(N, val)) for val in n_vals]

plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
fig, ax = plt.subplots(figsize=(11, 6.5), dpi=150)

ax.plot(n_vals, c_vals, label='Custo Total: C(N) = 46.750 + 535N', color='#E63946', lw=2.5)
ax.plot(n_vals, r_vals, label='Receita Total: R(N) = 785N', color='#2A9D8F', lw=2.5)
ax.plot(n_vals, l_vals, label='Lucro Líquido: L(N) = 250N - 46.750', color='#457B9D', lw=2, linestyle='--')

ax.scatter([float(N_breakeven)], [float(valor_breakeven)], color='#D90429', s=130, zorder=5)
ax.axvline(x=float(N_breakeven), color='gray', linestyle=':', alpha=0.7)
ax.axhline(y=float(valor_breakeven), color='gray', linestyle=':', alpha=0.7)
ax.axhline(y=0, color='black', lw=1, alpha=0.5)

ax.fill_between(n_vals, c_vals, r_vals, where=(n_vals < float(N_breakeven)), color='#E63946', alpha=0.15, label='Zona de Prejuízo')
ax.fill_between(n_vals, r_vals, c_vals, where=(n_vals >= float(N_breakeven)), color='#2A9D8F', alpha=0.15, label='Zona de Lucro Operacional')

ax.annotate(f'Ponto de Equilíbrio (Break-Even)\nN* = {int(N_breakeven)} robôs | R$ {float(valor_breakeven):,.2f}',
            xy=(float(N_breakeven), float(valor_breakeven)), xytext=(float(N_breakeven) - 110, float(valor_breakeven) + 40000),
            arrowprops=dict(facecolor='black', shrink=0.08, width=1.5, headwidth=8),
            fontsize=11, fontweight='bold', bbox=dict(boxstyle='round,pad=0.5', facecolor='#F1FAEE', edgecolor='#457B9D'))

ax.set_title('Hospital Alemão Oswaldo Cruz - Curvas de Custo, Receita e Lucro Operacional\nProjeto Robótica de Vistoria Hospitalar (FIAP)', fontsize=13, fontweight='bold', pad=15)
ax.set_xlabel('Quantidade de Robôs Produzidos e Vendidos (N)', fontsize=11, fontweight='bold')
ax.set_ylabel('Valores em Reais (R$)', fontsize=11, fontweight='bold')
ax.yaxis.set_major_formatter(ticker.StrMethodFormatter('R$ {x:,.0f}'))
ax.set_xlim(0, 350)
ax.set_ylim(-50000, 280000)
ax.legend(loc='upper left', frameon=True, facecolor='white', framealpha=0.95, fontsize=10)
plt.tight_layout()
plt.show()
print('Gráfico vetorial de alta resolução renderizado e salvo com sucesso!')
```

### 📊 Análise Executiva e Conclusões da Engenharia

1. **Viabilidade Operacional**:  
   - Produzindo e vendendo menos de **187 robôs**, a startup opera em **prejuízo** devido ao peso do custo fixo de infraestrutura (R$ 46.750,00).
   - A partir de **188 robôs**, cada unidade adicional contribui com **R$ 250,00 de lucro líquido direto**.

2. **Simulação de Meta para o Hospital**:  
   - Se o HAOC demandar uma escala de **250 robôs**, o faturamento bruto será de **R$ 196.250,00**, gerando um lucro líquido de **R$ 15.750,00/mês**.
   - Para uma escala de **300 robôs**, o lucro líquido salta para **R$ 28.250,00/mês** (margem líquida de ~12%).

3. **Recomendações Estratégicas**:  
   - Reduzir o custo unitário dos componentes eletrônicos ($C_u$) em 10% (de R$ 535 para R$ 481,50) reduziria o break-even para **154 robôs**, antecipando a lucratividade do projeto.
