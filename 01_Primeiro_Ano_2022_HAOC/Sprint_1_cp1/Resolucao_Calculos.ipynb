## Desafio Corporativo: Hospital Alemão Oswaldo Cruz

> **Especificação da Sprint:**
> Dimensionar a viabilidade econômico-financeira para implantação de uma frota de 4 robôs autônomos de transporte de medicamentos no Hospital Alemão Oswaldo Cruz. O investimento inicial de hardware e sensores é CAPEX = R$ 180.000,00 e custo operacional de energia/manutenção OPEX = R$ 2.400,00/mês. A operação manual atual demanda 6 maqueiros/mensageiros com custo mensal de R$ 4.200,00 cada. Determinar o Ponto de Equilíbrio (Break-Even Point em meses) e o VPL em 3 anos a uma TMA de 10% a.a.

### Modelagem Matemática Formal
$$\text{Break-Even} = \frac{\text{CAPEX}}{\Delta\text{OPEX}_{\text{mensal}}}, \quad \text{VPL} = -I_0 + \sum_{t=1}^{3} \frac{\text{Fluxo}_t}{(1 + \text{TMA})^t}$$


### Passo 1: Custos Mensais da Operação Manual Atual
Para 6 profissionais: Custo mensal = 6 \times R\$ 4.200,00 = R\$ 25.200,00/mês. Custo anual = 12 \times 25.200 = R\$ 302.400,00/ano.


### Passo 2: Custos da Operação Automatizada por Robôs
Custo operacional robôs = R\$ 2.400,00/mês. Economia líquida mensal gerada: \Delta\text{OPEX} = 25.200 - 2.400 = R\$ 22.800,00/mês (\approx R\$ 273.600,00/ano).


### Passo 3: Determinação do Ponto de Equilíbrio (Payback Simples)
\text{Break-Even} = \frac{180.000}{22.800} \approx 7,89 \text{ meses}. O investimento se paga totalmente em menos de 8 meses de operação contínua.


### Passo 4: Cálculo do Valor Presente Líquido (VPL a 10% a.a.)
Fluxo líquido anual = R\$ 273.600,00. Fator de anuidade para n=3 e i=10%: a_{3\rceil 10\%} = \frac{1 - (1,10)^{-3}}{0,10} \approx 2,48685. VPL = -180.000 + (273.600 \times 2,48685) = -180.000 + 680.402 = R\$ 500.402,00.


### Passo 5: Conclusão de Engenharia e Saúde
O projeto proporciona retorno expressivo sobre o capital empregado e libera a equipe de enfermagem para o atendimento clínico humanizado aos pacientes.


```python
import numpy as np
import matplotlib.pyplot as plt

meses = np.arange(0, 25)
capex = 180000
custo_manual = 25200 * meses
custo_robo = capex + 2400 * meses
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#0c233c')
ax.set_facecolor('#081626')
ax.plot(meses, custo_manual / 1000, color='#ef4444', lw=2, label='Custo Manual (6 Mensageiros)')
ax.plot(meses, custo_robo / 1000, color='#38bdf8', lw=2.5, label='Frota Robótica (CAPEX + OPEX)')
be_m = capex / (25200 - 2400)
ax.axvline(be_m, color='#f59e0b', ls='--', lw=1.5, label=f'Break-Even: {be_m:.1f} meses')
ax.scatter([be_m], [(capex + 2400*be_m)/1000], color='#f59e0b', s=60, zorder=5)
ax.set_title('Hospital Alemão Oswaldo Cruz: Análise de Break-Even Financeiro', color='#ffffff', fontsize=10.5, fontweight='bold')
ax.set_xlabel('Meses de Operação', color='#94a3b8', fontsize=9)
ax.set_ylabel('Custo Total Acumulado (R$ Mil)', color='#94a3b8', fontsize=9)
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#1e4d82', alpha=0.5, ls=':')
ax.legend(facecolor='#14355a', edgecolor='#1e4d82', labelcolor='#ffffff', fontsize=8)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```
