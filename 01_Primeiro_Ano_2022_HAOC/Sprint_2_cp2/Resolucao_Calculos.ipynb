> **Especificação da Sprint:**
> Dimensionar o trem de força e o torque requerido nos motores de corrente contínua para um robô hospitalar móvel com tração diferencial de massa total M = 60 kg (estrutura + carga de soros/medicamentos). O robô deve acelerar de 0 a v_max = 1,2 m/s em t = 1,5 s superando rampa hospitalar de inclinação θ = 5° com coeficiente de atrito de rolamento μ_r = 0,02 e raio das rodas R = 0,08 m.

### Modelagem Matemática Formal
$$F_{\text{total}} = M \cdot a + M \cdot g \cdot \sin\theta + M \cdot g \cdot \mu_r \cdot \cos\theta, \quad \tau_{\text{motor}} = \frac{F_{\text{total}} \cdot R}{2 \cdot \eta}$$


### Passo 1: Aceleração Requerida do Robô
Aceleração linear em linha reta: a = \frac{\Delta v}{\Delta t} = \frac{1,2}{1,5} = 0,8\text{ m/s}^2.


### Passo 2: Forças Resistivas na Rampa de 5 Graus
Com M = 60 kg e g = 9,81 m/s²: Força de inércia: F_i = 60 \times 0,8 = 48\text{ N}. Força gravitacional na rampa: F_g = 60 \times 9,81 \times \sin(5^\circ) = 588,6 \times 0,08715 \approx 51,29\text{ N}. Força de atrito de rolamento: F_r = 60 \times 9,81 \times 0,02 \times \cos(5^\circ) = 11,77 \times 0,996 \approx 11,73\text{ N}.


### Passo 3: Força Total de Tração Requerida
F_{total} = F_i + F_g + F_r = 48 + 51,29 + 11,73 = 111,02\text{ N}.


### Passo 4: Cálculo do Torque por Roda Motriz
Dividindo a tração entre os 2 motores com redutor de rendimento \eta = 85%: \tau = \frac{111,02 \times 0,08}{2 \times 0,85} = \frac{8,8816}{1,70} \approx 5,224\text{ N}\cdot\text{m} por motor.


### Passo 5: Especificação dos Motoredutores DC
Seleção de motor DC Brushless de 24 V, 150 W com caixa de redução planetária de 30:1, garantindo torque de pico de 8,5 N·m e margem de segurança de 62%.


```python
import numpy as np
import matplotlib.pyplot as plt

rampas_deg = np.linspace(0, 10, 100)
rampas_rad = np.radians(rampas_deg)
m, g, a, mu_r, r, eta = 60, 9.81, 0.8, 0.02, 0.08, 0.85
f_tot = m*a + m*g*np.sin(rampas_rad) + m*g*mu_r*np.cos(rampas_rad)
tau_mot = (f_tot * r) / (2 * eta)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#0c233c')
ax.set_facecolor('#081626')
ax.plot(rampas_deg, tau_mot, color='#38bdf8', lw=2.5, label='Torque por Motor (Nm)')
ax.axvline(5.0, color='#f59e0b', ls='--', label='Rampa Crítica Hospitalar (5° -> 5.22 Nm)')
ax.axhline(8.5, color='#ef4444', ls=':', lw=1.5, label='Torque Máximo Nominal do Motor (8.5 Nm)')
ax.set_title('Torque Requerido nos Motores vs. Inclinação da Rampa', color='#ffffff', fontsize=10.5, fontweight='bold')
ax.set_xlabel('Inclinação da Rampa (Graus)', color='#94a3b8', fontsize=9)
ax.set_ylabel('Torque por Motor (N·m)', color='#94a3b8', fontsize=9)
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#1e4d82', alpha=0.5, ls=':')
ax.legend(facecolor='#14355a', edgecolor='#1e4d82', labelcolor='#ffffff', fontsize=8)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```
