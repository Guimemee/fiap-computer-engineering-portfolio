> **Especificação da Sprint:**
> Projetar o circuito de instrumentação biomédica para captura de sinais de Eletromiografia de Superfície (sEMG) com amplificador de instrumentação de alto CMRR (> 110 dB). Implementar a filtragem passa-banda analógica de 20 a 450 Hz, retificação de onda completa, cálculo de envelope RMS e predição de intenção de movimento com Filtro de Kalman discreto executando sob o kernel FreeRTOS a 200 Hz.

### Modelagem Matemática Formal
$$\text{RMS}(t) = \sqrt{\frac{1}{W} \int_{t-W}^{t} s^2(\tau) d\tau}, \quad \text{Latência Intenção} \le 15 \text{ ms}$$


### Passo 1: Condicionamento de Sinais Biopotenciais EMG
Amplificação com ganho ajustável de 1.000x e rejeição de modo comum de ruído de rede elétrica (filtro Notch em 60 Hz).


### Passo 2: Extração de Envelope com Janela Deslizante de 50 ms
O sinal retificado gera curva de ativação proporcional à força de contração das fibras musculares.


### Passo 3: Algoritmo de Estimativa de Intenção com Filtro de Kalman
Eliminação de artefatos de movimento e estimativa suave do comando de velocidade angular desejado pelo operador.


### Passo 4: Sincronismo Temporal no FreeRTOS
A tarefa de leitura analógica roda com Prioridade Máxima a cada 5 ms cravados sem desvio de tick timer.


### Passo 5: Conclusão de Fluidez
Tempo entre o disparo neural muscular e a resposta mecânica do motor de apenas 11,8 ms, proporcionando sensação de transparência mecânica perfeita.


```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 4, 400)
# Sinal simulado de EMG durante contracao
raw_emg = np.where((t > 1.0) & (t < 3.0), np.random.RandomState(42).normal(0, 1.8, 400), np.random.RandomState(42).normal(0, 0.1, 400))
# Envelope RMS
envelope = np.where((t > 1.0) & (t < 3.0), 1.5 * np.sin(np.pi * (t - 1.0) / 2.0), 0.05)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#2d121e')
ax.set_facecolor('#1b0a12')
ax.plot(t, raw_emg, color='#94a3b8', lw=0.6, alpha=0.5, label='Sinal Bruto sEMG (μV)')
ax.plot(t, envelope * 2.5, color='#fb7185', lw=2.5, label='Envelope RMS Filtrado (Filtro Kalman)')
ax.set_title('Processamento de Sinal EMG e Predição de Intenção Motora', color='#ffffff', fontsize=10.5, fontweight='bold')
ax.set_xlabel('Tempo (Segundos)', color='#94a3b8', fontsize=9)
ax.set_ylabel('Amplitude Normalizada', color='#94a3b8', fontsize=9)
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#672d47', alpha=0.5, ls=':')
ax.legend(facecolor='#421c2d', edgecolor='#672d47', labelcolor='#ffffff', fontsize=8)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```
