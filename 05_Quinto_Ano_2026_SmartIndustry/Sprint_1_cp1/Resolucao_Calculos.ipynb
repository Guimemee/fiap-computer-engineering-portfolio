Dimensionar a estrutura cinemática em fibra de carbono e os redutores de velocidade de engrenagem harmônica (Harmonic Drive) para entrega de torque assistivo de até 22 N·m com peso estrutural inferior a 2,2 kg acoplado ao corpo.

### Modelagem Matemática Formal
$$\tau_{\text{assist}} = \beta \cdot \tau_{\text{biomecânico}}, \quad \beta \in [0,70; 0,85], \quad \sigma_{\text{vonMises}} \le \frac{S_y}{\text{FS}}$$


### Passo 1: Análise das Solicitações Musculares do Bíceps e Deltóide
Para um operador segurando ferramenta de 5,0 kg a 90 graus de flexão, o momento estático ultrapassa 20 N·m gerando fadiga aguda em menos de 4 minutos.


### Passo 2: Razão de Assistência e Ergonomia Ativa
Definição da taxa de suporte \beta = 75%: O exoesqueleto fornece 15,2 N·m, restando apenas 5,06 N·m ao músculo biológico.


### Passo 3: Dimensionamento Estrutural em Fibra de Carbono e Alumínio 7075-T6
Tensão de escoamento S_y = 503 MPa. Tensão máxima calculada de von Mises \sigma = 142 MPa, assegurando Fator de Segurança FS = 3,54.


### Passo 4: Seleção de Redutores Harmonic Drive de Zero Folga
Relação de redução de 50:1 com folga angular (backlash) menor que 1 arcmin, evitando trancos na articulação humana.


### Passo 5: Conclusão de Conforto e Segurança Biomecânica
O protótipo garante alívio imediato sem restringir a amplitude natural de movimento do braço (ROM de 0 a 140 graus).


```python
import numpy as np
import matplotlib.pyplot as plt

angulo = np.linspace(0, 140, 100)
ang_rad = np.radians(angulo)
tau_total = 20.26 * np.sin(ang_rad)
tau_assist = tau_total * 0.75
tau_usuario = tau_total * 0.25
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#2d121e')
ax.set_facecolor('#1b0a12')
ax.plot(angulo, tau_total, color='#ef4444', lw=2, label='Torque Biomecânico Total (Sem Exo)')
ax.plot(angulo, tau_assist, color='#fb7185', lw=2.5, label='Torque Assistivo do Motor (75%)')
ax.plot(angulo, tau_usuario, color='#10b981', lw=2, label='Esforço Remanescente do Usuário (25%)')
ax.set_title('Perfil de Torque Assistivo do Exoesqueleto Upper Limb', color='#ffffff', fontsize=10.5, fontweight='bold')
ax.set_xlabel('Ângulo de Flexão do Cotovelo (Graus)', color='#94a3b8', fontsize=9)
ax.set_ylabel('Torque Articular (N·m)', color='#94a3b8', fontsize=9)
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#672d47', alpha=0.5, ls=':')
ax.legend(facecolor='#421c2d', edgecolor='#672d47', labelcolor='#ffffff', fontsize=8)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```
